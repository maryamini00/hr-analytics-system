### hire date

```python
d['birth_date'] = pd.to_datetime(d['birth_date'])
today = "2026-08-01"
hd = np.empty(number_of_employees + 1, dtype='datetime64[D]')

for i in range(number_of_employees + 1):
    if d.iloc[i].loc["education_level"] == "کاردانی":
        gy = str(d.iloc[i].loc["graduation_year"]) + "-01-01"
        hd_possible = pd.date_range(start=gy, end=today)
        hd_choosed = np.random.choice(hd_possible)
        hd[i] = hd_choosed
    else:
        start_year = str(int(d.iloc[i].loc["birth_date"].year) + 24)
        start_month = str(d.iloc[i].loc["birth_date"].month)
        if int(start_month) < 10:
            start_month = "0" + start_month
        start_day = str(d.iloc[i].loc["birth_date"].day)
        if int(start_day) < 10:
            start_day = "0" + start_day
        start = start_year + '-' + start_month + '-' + start_day
        hd_possible = pd.date_range(start=start, end=today)
        n = len(hd_possible)
        weights = np.exp(-np.linspace(0, 3, n)) 
        weights /= weights.sum()
        hd_choosed = np.random.choice(hd_possible)
        hd[i] = hd_choosed
        
d["hire_date"] = hd   
```
